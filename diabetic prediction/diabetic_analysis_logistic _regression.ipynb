```python

col_names = ['pregnant', 'glucose', 'bp', 'skin', 'insulin', 'bmi', 'pedigree', 'age', 'label']
```

```python
dataset=pd.read_csv(r'C:\Users\SHUBHAM\OneDrive\Desktop\diabetic prediction/diabetes.csv',header=None, names=col_names)
```

```python
dataset
```

# eliminiate first rows

```python
dataset=dataset.iloc[1:]
```

```python
dataset.head()
```

```python
dataset.tail()
```

# spliat the dataset in feture and target variable

```python
feature_cols = ['pregnant', 'insulin', 'bmi', 'age','glucose','bp','pedigree']
```

```python
x=dataset[feature_cols]
y=dataset.label
```

```python
x
```

```python
y
```

# split x and y train and test 

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression(max_iter=1000)
```

```python
lr.fit(x_train,y_train)
```

```python
lr
```

# predicting the output for out test set

```python
y_pred=lr.predict(x_test)
```

```python
y_pred
```

```python
x_test
```

# confusion matrix

```python
from sklearn import metrics
cf_matrix=metrics.confusion_matrix(y_test,y_pred)
```

```python
cf_matrix
```

# ploting the confusion matrics

```python
cl_name=[0,1]
fix,ax=plt.subplots()
tick_marks=np.arange(len(cl_name))
plt.xticks(tick_marks,cl_name)
plt.yticks(tick_marks,cl_name)

# create heatmap
sns.heatmap(pd.DataFrame(cf_matrix),annot=True, cmap="YlGnBu",fmt='g')
ax.xaxis.set_label_position('top')
plt.tight_layout()
plt.title('Confusion matrix',y=1.1)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
```

```python
print('Accuracy:',metrics.accuracy_score(y_test,y_pred))
```
